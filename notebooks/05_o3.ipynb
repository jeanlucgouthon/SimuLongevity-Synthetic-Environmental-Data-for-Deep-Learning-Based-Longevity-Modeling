```python
import torch

import sys
from pathlib import Path
import matplotlib.pyplot as plt

PROJECT_ROOT = Path.cwd().parent
sys.path.insert(0,str(PROJECT_ROOT))
from src.simudata.o3 import simulate_o3
from src.visualizations.time_series_visualizer import plot_curve

from src.config import load_config
```

#### 1. Load temperature, humidity data

```python
temperature = torch.load(PROJECT_ROOT/"data/simulated_temperature.pth")
humidity = torch.load(PROJECT_ROOT/"data/simulated_humidity.pth")
```

#### 2. Simulation of 10,000 $O_3$ timesteps across 200 scenarios

```python
config = load_config(PROJECT_ROOT/"configs/simulations.yaml")
```

```python
humidity.shape
```

```python
R = 200
o3_data = torch.stack([simulate_o3(temperature_data=temperature[i],
                                       humidity_data=humidity[i],
                                        **config['o3']) for i in range(R)], dim=0)
```

#### 3. $O_3$ sample curve

```python
t= torch.arange(len(o3_data[0]))
plot_curve(t, o3_data[0,:2000],
           xlabel="timestep",
           ylabel="o3")
```

#### 4. Saving

```python
torch.save(o3_data, PROJECT_ROOT/"data/simulated_o3.pth")
```
