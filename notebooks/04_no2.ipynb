```python
import torch

import sys
from pathlib import Path
import matplotlib.pyplot as plt

PROJECT_ROOT = Path.cwd().parent
sys.path.insert(0,str(PROJECT_ROOT))
from src.simudata.no2 import simulate_no2
from src.visualizations.time_series_visualizer import plot_curve

from src.config import load_config
```

#### 1. Load temperature, humidity data

```python
temperature = torch.load(PROJECT_ROOT/"data/simulated_temperature.pth")
humidity = torch.load(PROJECT_ROOT/"data/simulated_humidity.pth")
```

#### 2. Simulation of 10,000 $NO_2$ timesteps across 200 scenarios

```python
config = load_config(PROJECT_ROOT/"configs/simulations.yaml")
```

```python
humidity.shape
```

```python
R = 200
no2_data = torch.stack([simulate_no2(temperature_data=temperature[i],
                                       humidity_data=humidity[i],
                                        **config['no2']) for i in range(R)], dim=0)
```

#### 3. $NO_2$ sample curve

```python
t= torch.arange(len(no2_data[0]))
plot_curve(t, no2_data[3,:2000],
           xlabel="timestep",
           ylabel="no2")
```

#### 4. Save $NO_2$ data

```python
torch.save(no2_data, PROJECT_ROOT/"data/simulated_no2.pth")
```
